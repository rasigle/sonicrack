# Creating sine wave

```python
figsize = (20, 6.25)
colors = "#323031", "#308E91", "#34369D", "#5E2A7E", "#5E2A7E", "#6F3384"
```

```python
import matplotlib.pyplot as plt

from engine import SineOscillator

plt.subplots(figsize=figsize)
plt.title("Single sine wave")

osc1 = SineOscillator(frequency=1, sample_rate=512)
samples = osc1.get_samples(n=512)

plt.ylabel("Amplitude")
plt.xlabel("Number of samples")
plt.plot(samples, color=colors[0])
plt.show()
```

## Sine multiple frequencies

```python
plt.subplots(figsize=figsize)
plt.title("Different Frequencies")

for hz, c in zip([1, 2, 3, 4], colors, strict=False):
    osc = SineOscillator(frequency=hz)
    wave = osc.get_samples()
    plt.plot(wave, label=f"{hz} Hz", color=c)

plt.ylabel("Amplitude")
plt.xlabel("Number of samples")
plt.legend(loc="lower right")
plt.show()
```

## Different amplitudes

```python
plt.subplots(figsize=figsize)
plt.title("Different Amplitudes")

for amp, c in zip([1, 0.75, 0.5, 0.25], colors, strict=False):
    osc1 = SineOscillator(frequency=2, amplitude=amp)
    wave = osc1.get_samples()
    plt.plot(wave, label=f"amplitude={amp}", color=c)

plt.ylabel("Amplitude")
plt.xlabel("Number of samples")
plt.legend(loc="lower right")
plt.show()
```

## Different phases

```python
plt.subplots(figsize=figsize)
plt.title("Different Phases")

for phase, c in zip([0, 45, 90, 135], colors, strict=False):
    osc1 = SineOscillator(frequency=2, phase=phase)
    wave = osc1.get_samples()
    plt.plot(wave, label=f"phase={phase}°", color=c)

plt.ylabel("Amplitude")
plt.xlabel("Number of samples")
plt.legend(loc="lower right")
plt.show()
```
