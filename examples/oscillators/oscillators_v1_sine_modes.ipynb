# Creating sine wave

```python
figsize = (20., 6.25)
colors = "#323031", "#308E91", "#34369D", "#5E2A7E", "#5E2A7E", "#6F3384"
```

```python
import matplotlib.pyplot as plt

from engine import SineOscillator

plt.figure(figsize=figsize)
plt.title("Single sine wave")

sine1 = SineOscillator(frequency=1, sample_rate=512, mode="pure", gain_db=0)
samples1 = sine1.get_samples(n=512)

sine2 = SineOscillator(frequency=1, sample_rate=512, mode="bright", gain_db=0)
samples2 = sine2.get_samples(n=512)

sine3 = SineOscillator(frequency=1, sample_rate=512, mode="warm", gain_db=0)
samples3 = sine3.get_samples(n=512)

sine4 = SineOscillator(frequency=1, sample_rate=512, mode="analog", gain_db=0)
samples4 = sine4.get_samples(n=512)

sine5 = SineOscillator(frequency=1, sample_rate=512, mode="vcv", gain_db=0)
samples5 = sine5.get_samples(n=512)

plt.ylabel("Amplitude")
plt.xlabel("Number of samples")
plt.plot(samples1, color=colors[0], label="Pure")
plt.plot(samples2, color=colors[1], label="Bright")
plt.plot(samples3, color=colors[2], label="Warm")
plt.plot(samples4, color=colors[2], label="Analog")
plt.plot(samples5, color=colors[3], label="VCV")
plt.legend()
plt.show()
```
