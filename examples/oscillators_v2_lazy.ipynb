# Creating sine wave

```python
import matplotlib.pyplot as plt

from engine.oscillators import SineOscillator
from src.engine.generators_lazy import Sine

# Create a sine wave oscillator
osc = Sine(freq=1, amp=1, phase_deg=0, sample_rate=512)
osc2 = SineOscillator(freq=1, amp=1, phase=0, sample_rate=512)

# Get 512 samples from the oscillator
samples1_1 = osc.get_samples(n=512)
osc.amp = 2
osc.freq = 10
samples1_2 = osc.get_samples(n=512)
samples2_1 = osc2.get_samples(n=512)
osc2.amp = 2
osc2.freq = 5
osc2.phase = 0
samples2_2 = osc2.get_samples(n=512)

plt.plot(samples1_1+samples1_2)
plt.plot(samples2_1+samples2_2)
plt.show()
```
