```python
import matplotlib.pyplot as plt
from engine.oscillators import (
    SineOscillator,
    SquareOscillator,
    SawtoothOscillator,
    TriangleOscillator,
)
from engine.composers import WaveAdder

# Create a sine wave oscillator
osc1 = SineOscillator(freq=1, amp=1, phase=0, sample_rate=512)
osc2 = SquareOscillator(freq=2, amp=0.5, phase=45, sample_rate=512)

adder = WaveAdder(osc1, osc2)

# Get 512 samples from the oscillator
samples = adder.get_samples(n=512 * 5)
plt.plot(samples)
plt.show()
```

```python
import matplotlib.pyplot as plt
from engine.oscillators import (
    SineOscillator,
    SquareOscillator,
    SawtoothOscillator,
    TriangleOscillator,
)
from engine.composers import Chain

# Create a sine wave oscillator
osc1 = SineOscillator(freq=1, amp=1, phase=0, sample_rate=512)
osc2 = SquareOscillator(freq=2, amp=0.5, phase=45, sample_rate=512)

chain = Chain(osc1, osc2)

# Get 512 samples from the oscillator
samples = chain.get_samples(n=512 * 5)
plt.plot(samples)
plt.show()
```
